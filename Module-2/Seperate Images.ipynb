```python
from deepface import DeepFace
import os
import shutil
```

```python


input_folder = "dataset/mixed_faces"
male_folder = "dataset/male_faces"
female_folder = "dataset/female_faces"

os.makedirs(male_folder, exist_ok=True)
os.makedirs(female_folder, exist_ok=True)

for img_name in os.listdir(input_folder):
    img_path = os.path.join(input_folder, img_name)

    try:
        # Predict gender
        analysis = DeepFace.analyze(img_path, actions=['gender'], enforce_detection=False)
        gender = analysis[0]['dominant_gender']

        # Move image to respective folder
        if gender in ["Man", "male"]:
            shutil.move(img_path, os.path.join(male_folder, img_name))
        else:
            shutil.move(img_path, os.path.join(female_folder, img_name))

    except Exception as e:
        print(f"Error processing {img_name}: {e}")

print("Sorting completed successfully!")
```
